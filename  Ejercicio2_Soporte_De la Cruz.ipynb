## Ejercicio 2: Escalamiento de tickets de soporte técnico

### Ficha PEAS

| Elemento | Descripción |
|---|---|
| **Percepción (S)** | tiempo_espera_minutos (entero), nivel_urgencia (baja/media/alta), cliente_premium (True/False) |
| **Acciones (A)** | escalar a nivel 1, escalar a nivel 2, escalar a nivel 3, mantener en nivel 1 |
| **Entorno (E)** | Mesa de ayuda de soporte técnico; tickets entrantes generados por clientes en tiempo real |
| **Objetivo** | Atender cada ticket en un tiempo adecuado a su urgencia y tipo de cliente, priorizando los casos críticos |
| **Medida de desempeño (P)** | Tiempo promedio de resolución por nivel de urgencia, cumplimiento del SLA de clientes premium, número de tickets urgentes atendidos a tiempo |

### Justificación de las reglas

Un ticket de urgencia alta se atiende de inmediato porque un problema grave empeora si se deja esperando, sin importar quién lo reportó. Los clientes premium tienen prioridad porque pagan por un servicio más rápido, así que incluso con problemas menores se les atiende antes. El tiempo de espera se usa para los casos de urgencia media, para que no se queden esperando demasiado tiempo aunque no sean premium.
