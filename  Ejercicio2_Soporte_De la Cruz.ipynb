## Ejercicio 2: Escalamiento de tickets de soporte técnico

### Ficha PEAS

| Elemento | Descripción |
|---|---|
| **Percepción (S)** | tiempo_espera_minutos (entero), nivel_urgencia (baja/media/alta), cliente_premium (True/False) |
| **Acciones (A)** | escalar a nivel 1, escalar a nivel 2, escalar a nivel 3, mantener en nivel 1 |
| **Entorno (E)** | Mesa de ayuda de soporte técnico; tickets entrantes generados por clientes en tiempo real |
| **Objetivo** | Atender cada ticket en un tiempo adecuado a su urgencia y tipo de cliente, priorizando los casos críticos |
| **Medida de desempeño (P)** | Tiempo promedio de resolución por nivel de urgencia, cumplimiento del SLA de clientes premium, número de tickets urgentes atendidos a tiempo |

### Justificación de las reglas

Un ticket de alta urgencia se atiende de inmediato porque un problema grave empeora si se deja esperando, sin importar quién lo reportó. Los clientes premium tienen prioridad porque pagan por un servicio más rápido, así que incluso con problemas menores se les atiende antes. El tiempo de espera se usa para los casos de urgencia media, para que no se queden esperando demasiado tiempo aunque no sean premium.

### Código

```python
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):
    # Define la funcion que recibe los 3 datos que percibe el agente 

    if nivel_urgencia == "alta":
        # Si la urgencia es alta, se escala de inmediato
        return "escalar a nivel 3", "urgencia alta reportada por el cliente"

    elif nivel_urgencia == "media":
        # Si no era alta, revisa si es "media"

        if tiempo_espera_minutos >= 20:
            # Dentro de urgencia media, si ya espero 20 minutos o mas, se escala
            return "escalar a nivel 2", "urgencia media con tiempo de espera >= 20 minutos"

        elif cliente_premium:
            # Si espero menos de 20 min, pero es cliente premium, tambien se escala
            return "escalar a nivel 2", "urgencia media y cliente premium"

        else:
            # Urgencia media, poco tiempo de espera y no es premium: se queda en nivel 1
            return "mantener en nivel 1", "urgencia media, tiempo de espera bajo y sin prioridad premium"

    else:
        # Si no fue "alta" ni "media", entonces nivel_urgencia es "baja"

        if cliente_premium and tiempo_espera_minutos >= 30:
            # Urgencia baja, pero es premium y ya espero mucho mas de 30 min, se escala igual
            return "escalar a nivel 1", "urgencia baja pero cliente premium con espera prolongada"

        else:
            # Cualquier otro caso de urgencia baja: atencion normal, sin escalar
            return "mantener en nivel 1", "urgencia baja, atencion estandar"
```

### Simulacion y pruebas

```python
casos = [
    # (tiempo_espera_minutos, nivel_urgencia, cliente_premium)
    (5, "alta", False),      
    (25, "media", False),    
    (10, "media", True),     
    (10, "media", False),    
    (35, "baja", True),      
    (35, "baja", False),     
    (20, "media", False),    
]
for tiempo, urgencia, premium in casos:
    accion, motivo = agente_soporte(tiempo, urgencia, premium)
    print(f"Tiempo={tiempo} min, Urgencia={urgencia}, Premium={premium}")
    print(f"Accion: {accion} | Motivo: {motivo}\n")
```
