## Ejercicio 3: Inspección visual de un producto 

### Ficha PEAS

| Elemento | Descripción |
|---|---|
| **Percepción (S)** | Matriz 5x5 de NumPy con valores 0 (píxel normal) o 1 (píxel defectuoso) |
| **Acciones (A)** | aprobar, revisión manual, rechazar |
| **Entorno (E)** | Línea de producción industrial; cada producto pasa una sola vez frente a la cámara |
| **Objetivo** | Minimizar los productos defectuosos que llegan al cliente sin descartar productos buenos innecesariamente |
| **Medida de desempeño (P)** | Tasa de defectos correctamente detectados, tasa de falsos rechazos, productividad de la línea (productos aprobados por hora) |

### Justificacion de las reglas

Contar los píxeles defectuosos es una forma simple de medir qué tan dañado está el producto. Si no hay ningún defecto, se aprueba directo. Si hay pocos defectos (1 a 3) no se rechaza automáticamente porque podría ser solo ruido de la cámara, así que se manda a revisión manual para que una persona lo confirme. Si hay más de 3 defectos ya es un daño claro y se rechaza sin necesidad de revisión.
